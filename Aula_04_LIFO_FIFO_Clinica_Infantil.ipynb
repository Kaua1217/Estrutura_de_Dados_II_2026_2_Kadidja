# Aula 04 — Estruturas LIFO e FIFO

## Fluxo de Preferência no Atendimento de uma Clínica Infantil

Nesta atividade serão aplicados os conceitos de **Pilha (LIFO)**, **Fila (FIFO)**, **Fila com preferência**, `deque`, **Pilha via Array Estático** e **Fila Circular**.

### Contexto

A clínica infantil possui uma fila de atendimento:

- Crianças que chegam normalmente entram no final da fila.
- O atendimento normal segue o conceito **FIFO (First In, First Out)**.
- Crianças com indicação de atendimento com preferência podem ser inseridas no início da fila.
- O atendimento retira o primeiro paciente da fila.
- Situações de histórico de alterações podem utilizar uma **Pilha (LIFO)**.


## Regras utilizadas no contexto

### Fila normal

```text
Ana → Pedro → Lucas → Maria
```

### Entrada de paciente com preferência

```text
João → Ana → Pedro → Lucas → Maria
```

A preferência será representada por:

```python
fila.appendleft(paciente)
```

O atendimento será realizado com:

```python
fila.popleft()
```


# 1. Exercício 1 — O Simulador "Desfazer"

## Objetivo

Utilizar uma **pilha** para armazenar ações e permitir que a última ação realizada seja a primeira a ser desfeita.

## Adaptação para a clínica

A pilha representa o histórico de alterações realizadas no fluxo de atendimento, como registrar chegada de paciente ou alterar uma preferência.

## Conceito

**Pilha — LIFO (Last In, First Out)**

- Inserção: `append()`
- Remoção: `pop()`


````python
# Pilha que armazenará o histórico de ações
historico = []

def registrar_acao(acao):
    historico.append(acao)
    print(f"Ação registrada: {acao}")

def desfazer():
    if len(historico) == 0:
        print("Não há ações para desfazer.")
    else:
        acao = historico.pop()
        print(f"Ação desfeita: {acao}")

def mostrar_historico():
    print("\nHistórico atual:")
    if len(historico) == 0:
        print("Nenhuma ação registrada.")
    else:
        for acao in historico:
            print("-", acao)
````

````python
# Teste do exercício
registrar_acao("Registrar chegada de Ana")
registrar_acao("Registrar chegada de Pedro")
registrar_acao("Alterar prioridade de Pedro")

mostrar_historico()

print("\n--- DESFAZENDO ÚLTIMA AÇÃO ---")
desfazer()
mostrar_historico()

print("\n--- DESFAZENDO NOVAMENTE ---")
desfazer()
mostrar_historico()
````

## Resultado

A última ação registrada é a primeira a ser removida.

Isso caracteriza o princípio **LIFO**:
**Last In, First Out**.


# 2. Exercício 2 — O Sistema de Impressão

## Objetivo

Utilizar uma **fila FIFO** para processar elementos na mesma ordem em que foram inseridos.

## Adaptação para a clínica

Os documentos de atendimento da clínica entram na fila e são processados na ordem de chegada.

## Conceito

**Fila — FIFO (First In, First Out)**

Com `deque`:

- Entrada no final: `append()`
- Remoção no início: `popleft()`


````python
from collections import deque

# Fila de documentos da clínica
fila_documentos = deque()

def adicionar_documento(documento):
    fila_documentos.append(documento)
    print(f"Documento adicionado: {documento}")
    print(f"Fila atual: {list(fila_documentos)}")

def processar_documento():
    if len(fila_documentos) == 0:
        print("Não existem documentos aguardando.")
    else:
        documento = fila_documentos.popleft()
        print(f"Documento processado: {documento}")
        print(f"Fila atual: {list(fila_documentos)}")
````

````python
# Teste do exercício
adicionar_documento("Ficha de Ana")
adicionar_documento("Ficha de Pedro")
adicionar_documento("Ficha de Lucas")
adicionar_documento("Ficha de Maria")

print("\n--- PROCESSAMENTO ---")
processar_documento()
processar_documento()
processar_documento()
processar_documento()
````

## Resultado

A ordem de processamento será:

```text
Ana → Pedro → Lucas → Maria
```

O primeiro documento inserido é o primeiro a ser processado. Portanto, é **FIFO**.


# 3. Desafio Master — Triagem Hospitalar

## Objetivo

Simular uma fila de atendimento com pacientes normais e pacientes que possuem preferência.

## Adaptação para a clínica

No contexto da clínica infantil:

- Pacientes normais entram no final da fila.
- Pacientes com preferência entram no início da fila.
- O atendimento sempre retira o paciente que está no início.

### Conceito

A estrutura utilizada será `collections.deque`.

```python
fila.append(paciente)       # entrada normal
fila.appendleft(paciente)   # entrada com preferência
fila.popleft()              # atendimento
```

A preferência é uma **regra de inserção** dentro de uma fila; o atendimento continua ocorrendo pelo início.


````python
from collections import deque

# Fila de atendimento da clínica
fila = deque()

def mostrar_fila():
    if len(fila) == 0:
        print("Fila vazia.")
    else:
        print("Fila de atendimento:")
        for i, paciente in enumerate(fila, start=1):
            print(f"{i}. {paciente}")

def adicionar_paciente(nome, preferencia=False):
    if preferencia:
        fila.appendleft(nome)
        print(f"\n{nome} chegou com PREFERÊNCIA.")
        print("Paciente inserido no início da fila.")
    else:
        fila.append(nome)
        print(f"\n{nome} chegou para atendimento NORMAL.")
        print("Paciente inserido no final da fila.")

    mostrar_fila()

def atender_paciente():
    if len(fila) == 0:
        print("\nNão existem pacientes aguardando.")
    else:
        paciente = fila.popleft()
        print(f"\nPaciente atendido: {paciente}")
        print("Fila após o atendimento:")
        mostrar_fila()
````

````python
# 1. Chegada de pacientes normais
adicionar_paciente("Ana")
adicionar_paciente("Pedro")
adicionar_paciente("Lucas")
adicionar_paciente("Maria")

# 2. Chegada de paciente com preferência
adicionar_paciente("João", preferencia=True)

# 3. Atendimento dos pacientes
print("\n=== INÍCIO DOS ATENDIMENTOS ===")
atender_paciente()
atender_paciente()
atender_paciente()
atender_paciente()
atender_paciente()
````

## Resultado do Desafio Master

Depois das chegadas normais:

```text
Ana → Pedro → Lucas → Maria
```

Quando João chega com preferência:

```text
João → Ana → Pedro → Lucas → Maria
```

A ordem de atendimento será:

```text
1. João
2. Ana
3. Pedro
4. Lucas
5. Maria
```

A preferência foi aplicada na **entrada**, com `appendleft()`, enquanto a remoção continua sendo feita com `popleft()`.


# 4. Exercício 01 — A Pilha Matemática

## Objetivo

Criar uma pilha de números inteiros e esvaziá-la iterativamente, calculando a soma dos elementos removidos.

## Adaptação para a clínica

Os números representam códigos de fichas dos pacientes. A soma é apenas a operação matemática solicitada pelo exercício.

## Conceito

**Pilha — LIFO**

- Inserção: `append()`
- Remoção: `pop()`


````python
# Pilha com códigos de fichas
pilha = []

pilha.append(101)  # Ana
pilha.append(102)  # Pedro
pilha.append(103)  # Lucas
pilha.append(104)  # Maria

print("Pilha inicial:")
print(pilha)

soma = 0

print("\nRemovendo elementos da pilha:")

while len(pilha) > 0:
    numero = pilha.pop()
    soma += numero
    print(f"Removido: {numero}")
    print(f"Soma atual: {soma}")

print("\nSoma total dos elementos removidos:", soma)
````

## Resultado

Como a pilha utiliza LIFO, a ordem de remoção será:

```text
104 → 103 → 102 → 101
```

A soma total será:

```text
410
```


# 5. Exercício 02 — O Call Center

## Objetivo

Criar uma fila FIFO em que novos clientes entram no final e o próximo cliente é chamado pelo início.

## Adaptação para a clínica

Os clientes serão representados por pacientes infantis.

## Conceito

**Fila — FIFO**

- Entrada: `append()`
- Atendimento: `popleft()`


````python
from collections import deque

fila = deque()

def adicionar_paciente(nome):
    fila.append(nome)
    print(f"\nPaciente chegou: {nome}")
    print("Fila:", list(fila))

def chamar_paciente():
    if len(fila) == 0:
        print("\nNão existem pacientes aguardando.")
    else:
        paciente = fila.popleft()
        print(f"\nChamando para atendimento: {paciente}")
        print("Fila restante:", list(fila))
````

````python
# Teste do exercício
adicionar_paciente("Ana")
adicionar_paciente("Pedro")
adicionar_paciente("Lucas")
adicionar_paciente("Maria")

print("\n=== ATENDIMENTO ===")
chamar_paciente()
chamar_paciente()
chamar_paciente()
chamar_paciente()
````

## Resultado

A ordem de chegada e atendimento será:

```text
Ana → Pedro → Lucas → Maria
```

O exercício demonstra **FIFO — First In, First Out**.


# 6. Exercício 01 — Pilha via Array Estático

## Objetivo

Implementar uma pilha utilizando um array de tamanho fixo e uma variável `top` para controlar o topo.

## Adaptação para a clínica

A pilha armazenará códigos de fichas dos pacientes.

A implementação abaixo não usa `append()` nem `pop()` para controlar a pilha, pois o objetivo é demonstrar o funcionamento do array estático e do `top`.

## Conceito

**Pilha — LIFO**

- `top = -1` representa pilha vazia.
- `top += 1` ao inserir.
- `top -= 1` ao remover.


````python
# Capacidade máxima da pilha
CAPACIDADE = 5

# Array de tamanho fixo
pilha = [None] * CAPACIDADE

# Topo da pilha
top = -1

def push(valor):
    global top

    if top == CAPACIDADE - 1:
        print("Erro: a pilha está cheia.")
        return

    top += 1
    pilha[top] = valor
    print(f"Ficha {valor} adicionada.")

def pop():
    global top

    if top == -1:
        print("Erro: a pilha está vazia.")
        return None

    valor = pilha[top]
    pilha[top] = None
    top -= 1

    print(f"Ficha {valor} removida.")
    return valor

def mostrar_pilha():
    print("\nPilha atual:")
    if top == -1:
        print("Pilha vazia.")
    else:
        for i in range(top, -1, -1):
            print(pilha[i])
````

````python
# Teste da pilha estática
push(101)
push(102)
push(103)
push(104)

mostrar_pilha()

print("\n--- REMOÇÕES ---")
pop()
pop()

mostrar_pilha()
````

## Resultado

As fichas são retiradas na ordem:

```text
104 → 103 → 102 → 101
```

O `top` controla a posição do último elemento inserido, mantendo o comportamento **LIFO**.


# 7. Exercício 02 — Fila Circular

## Objetivo

Implementar uma fila utilizando um array de tamanho fixo, controlando `front` e `rear`.

A fila é circular porque, ao chegar ao final do array, o próximo índice retorna para o início.

## Adaptação para a clínica

A fila circular representa a fila de pacientes da clínica infantil.

## Conceito

**Fila — FIFO**

- `front`: posição do primeiro paciente.
- `rear`: posição onde o próximo paciente será inserido.
- `quantidade`: quantidade atual de pacientes.
- Fórmula circular: `(indice + 1) % CAPACIDADE`.


````python
# Capacidade máxima da fila
CAPACIDADE = 5

# Array da fila
fila = [None] * CAPACIDADE

# Índice do primeiro elemento
front = 0

# Índice onde o próximo elemento será inserido
rear = 0

# Quantidade de elementos
quantidade = 0

def enqueue(paciente):
    global rear, quantidade

    if quantidade == CAPACIDADE:
        print("Erro: a fila está cheia.")
        return

    fila[rear] = paciente

    print(f"\nPaciente entrou na fila: {paciente}")

    rear = (rear + 1) % CAPACIDADE
    quantidade += 1

    mostrar_fila()

def dequeue():
    global front, quantidade

    if quantidade == 0:
        print("\nA fila está vazia.")
        return None

    paciente = fila[front]
    fila[front] = None

    print(f"\nPaciente atendido: {paciente}")

    front = (front + 1) % CAPACIDADE
    quantidade -= 1

    mostrar_fila()

    return paciente

def mostrar_fila():
    print("Array atual:", fila)
    print(f"front = {front}")
    print(f"rear = {rear}")
    print(f"quantidade = {quantidade}")
````

````python
# Reiniciando a fila circular
fila = [None] * CAPACIDADE
front = 0
rear = 0
quantidade = 0

print("=== CHEGADA DOS PACIENTES ===")

enqueue("Ana")
enqueue("Pedro")
enqueue("Lucas")
enqueue("Maria")

print("\n=== ATENDIMENTO ===")

dequeue()
dequeue()

print("\n=== NOVAS CHEGADAS ===")

enqueue("João")
enqueue("Mariana")

print("\n=== NOVO ATENDIMENTO ===")

dequeue()
dequeue()
````

## Resultado

A fila circular mantém a ordem FIFO mesmo reutilizando posições liberadas do array.

A sequência de atendimento começa por:

```text
Ana → Pedro
```

e depois continua com os pacientes que permanecem na fila.

Quando o índice chega ao final do array, a expressão:

```python
(rear + 1) % CAPACIDADE
```

faz o índice retornar ao início, permitindo o reaproveitamento das posições.
