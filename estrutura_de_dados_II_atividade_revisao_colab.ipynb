## 📝 Exercício 1 — Soma de um vetor

Leia 10 números inteiros e armazene-os em uma lista.

Ao final, apresente:
- todos os números;
- a soma dos elementos;
- a média dos valores.


````python
numeros = []

for i in range(10):
    numero = int(input(f"Digite o {i+1}º número: "))
    numeros.append(numero)

soma = sum(numeros)
media = soma / len(numeros)

print("\nNúmeros:", numeros)
print("Soma:", soma)
print(f"Média: {media:.2f}")
````

## 📝 Exercício 2 — Maior e menor

Leia 10 números inteiros utilizando uma lista.

Determine:
- o maior valor;
- o menor valor;
- as posições em que eles aparecem.


````python
numeros = []

for i in range(10):
    numeros.append(int(input(f"Digite o {i+1}º número: ")))

maior = max(numeros)
menor = min(numeros)

posicoes_maior = [i for i, valor in enumerate(numeros) if valor == maior]
posicoes_menor = [i for i, valor in enumerate(numeros) if valor == menor]

print("\nVetor:", numeros)
print("Maior valor:", maior)
print("Posições do maior:", posicoes_maior)
print("Menor valor:", menor)
print("Posições do menor:", posicoes_menor)
````

## 📝 Exercício 3 — Números pares

Leia 20 números inteiros e armazene-os em uma lista.

Depois:
- mostre somente os números pares;
- conte quantos números pares existem;
- calcule a soma dos números pares.


````python
numeros = []

for i in range(20):
    numeros.append(int(input(f"Digite o {i+1}º número: ")))

pares = [numero for numero in numeros if numero % 2 == 0]

print("\nNúmeros pares:", pares)
print("Quantidade de pares:", len(pares))
print("Soma dos pares:", sum(pares))
````

## 📝 Exercício 4 — Inversão de vetor

Leia 10 números e armazene-os em uma lista.

Mostre o vetor original e o vetor invertido.

**Desafio:** não utilize outra lista para realizar a inversão.


````python
numeros = []

for i in range(10):
    numeros.append(int(input(f"Digite o {i+1}º número: ")))

print("\nVetor original:", numeros)

# Inversão na própria lista
inicio = 0
fim = len(numeros) - 1

while inicio < fim:
    numeros[inicio], numeros[fim] = numeros[fim], numeros[inicio]
    inicio += 1
    fim -= 1

print("Vetor invertido:", numeros)
````

## 📝 Exercício 5 — Matriz 3 × 3

Crie uma matriz 3 × 3 de números inteiros.

O programa deverá:
- ler os valores;
- mostrar a matriz;
- calcular a soma de todos os elementos;
- mostrar o maior valor.


````python
matriz = []

for i in range(3):
    linha = []
    for j in range(3):
        linha.append(int(input(f"Digite o valor [{i}][{j}]: ")))
    matriz.append(linha)

print("\nMatriz:")
for linha in matriz:
    print(linha)

todos = [valor for linha in matriz for valor in linha]

print("\nSoma dos elementos:", sum(todos))
print("Maior valor:", max(todos))
````

## 📝 Exercício 6 — Diagonal principal

Leia uma matriz 4 × 4.

Mostre os elementos da diagonal principal e calcule sua soma.


````python
matriz = []

for i in range(4):
    linha = []
    for j in range(4):
        linha.append(int(input(f"Digite o valor [{i}][{j}]: ")))
    matriz.append(linha)

print("\nMatriz:")
for linha in matriz:
    print(linha)

diagonal = [matriz[i][i] for i in range(4)]

print("\nDiagonal principal:", diagonal)
print("Soma da diagonal:", sum(diagonal))
````

## 📝 Exercício 7 — Matriz de notas

Uma turma possui 4 alunos e 3 avaliações.

Crie uma matriz para armazenar as notas e calcule a média de cada aluno.


````python
notas = []

for aluno in range(4):
    notas_aluno = []
    print(f"\nAluno {aluno + 1}")

    for avaliacao in range(3):
        nota = float(input(f"Digite a nota {avaliacao + 1}: "))
        notas_aluno.append(nota)

    notas.append(notas_aluno)

print("\n=== MÉDIAS ===")

for i, aluno in enumerate(notas):
    media = sum(aluno) / len(aluno)
    print(f"Aluno {i + 1}: {media:.2f}")
````

## 📝 Exercício 8 — Struct Produto

Em C/C++, o exercício utiliza uma `struct Produto`. Em Python, podemos representar essa mesma ideia utilizando uma `dataclass`.

Crie uma estrutura contendo:
- nome;
- código;
- preço;
- quantidade.

Cadastre 5 produtos e:
- mostre todos os produtos;
- calcule o valor total de cada produto;
- informe o produto com maior valor em estoque.

**Fórmula:** `valor em estoque = preço × quantidade`


````python
from dataclasses import dataclass

@dataclass
class Produto:
    nome: str
    codigo: int
    preco: float
    quantidade: int

produtos = []

for i in range(5):
    print(f"\nProduto {i + 1}")

    nome = input("Nome: ")
    codigo = int(input("Código: "))
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade: "))

    produtos.append(Produto(nome, codigo, preco, quantidade))

print("\n=== PRODUTOS ===")

for produto in produtos:
    valor_estoque = produto.preco * produto.quantidade
    print(f"Nome: {produto.nome}")
    print(f"Código: {produto.codigo}")
    print(f"Preço: R$ {produto.preco:.2f}")
    print(f"Quantidade: {produto.quantidade}")
    print(f"Valor em estoque: R$ {valor_estoque:.2f}")
    print("-" * 30)

maior = max(produtos, key=lambda p: p.preco * p.quantidade)

print("\nProduto com maior valor em estoque:")
print(maior.nome)
print(f"Valor: R$ {maior.preco * maior.quantidade:.2f}")
````

## 📝 Exercício 9 — Struct Aluno

Em Python, utilize uma `dataclass` para representar a estrutura `Aluno`.

Cada aluno deve possuir:
- nome;
- idade;
- três notas.

Cadastre 5 alunos.

Classificação:
- média >= 7,0 → **Aprovado**
- média < 7,0 → **Reprovado**

Ao final, informe:
- quantidade de aprovados;
- quantidade de reprovados;
- aluno com maior média.


````python
from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    idade: int
    notas: list

alunos = []

for i in range(5):
    print(f"\nAluno {i + 1}")

    nome = input("Nome: ")
    idade = int(input("Idade: "))

    notas = []
    for j in range(3):
        notas.append(float(input(f"Nota {j + 1}: ")))

    alunos.append(Aluno(nome, idade, notas))

aprovados = 0
reprovados = 0

print("\n=== RESULTADO ===")

for aluno in alunos:
    media = sum(aluno.notas) / len(aluno.notas)

    if media >= 7:
        situacao = "Aprovado"
        aprovados += 1
    else:
        situacao = "Reprovado"
        reprovados += 1

    print(f"{aluno.nome} | Média: {media:.2f} | {situacao}")

maior = max(alunos, key=lambda a: sum(a.notas) / len(a.notas))
media_maior = sum(maior.notas) / len(maior.notas)

print("\nQuantidade de aprovados:", aprovados)
print("Quantidade de reprovados:", reprovados)
print(f"Aluno com maior média: {maior.nome} ({media_maior:.2f})")
````

# 📝 Exercício 10 — Sistema Integrado

Crie um pequeno sistema para cadastro de **10 funcionários**.

Cada funcionário deve possuir:
- nome;
- idade;
- cargo;
- salário.

O sistema deverá permitir:
1. Cadastrar funcionários.
2. Listar funcionários.
3. Mostrar o funcionário com maior salário.
4. Calcular a média salarial.
5. Mostrar funcionários com salário acima da média.

### Desafio extra

Implemente um menu:

```text
=================================
       SISTEMA DE FUNCIONÁRIOS
=================================

1 - Cadastrar funcionários
2 - Listar funcionários
3 - Maior salário
4 - Média salarial
5 - Salários acima da média
0 - Sair
```


````python
from dataclasses import dataclass

@dataclass
class Funcionario:
    nome: str
    idade: int
    cargo: str
    salario: float

funcionarios = []

def cadastrar():
    if len(funcionarios) >= 10:
        print("\nLimite de 10 funcionários atingido.")
        return

    print("\n=== CADASTRO DE FUNCIONÁRIO ===")
    nome = input("Nome: ")
    idade = int(input("Idade: "))
    cargo = input("Cargo: ")
    salario = float(input("Salário: R$ "))

    funcionarios.append(Funcionario(nome, idade, cargo, salario))
    print("Funcionário cadastrado com sucesso!")

def listar():
    if not funcionarios:
        print("\nNenhum funcionário cadastrado.")
        return

    print("\n=== FUNCIONÁRIOS ===")
    for i, funcionario in enumerate(funcionarios, start=1):
        print(f"\nFuncionário {i}")
        print(f"Nome: {funcionario.nome}")
        print(f"Idade: {funcionario.idade}")
        print(f"Cargo: {funcionario.cargo}")
        print(f"Salário: R$ {funcionario.salario:.2f}")

def maior_salario():
    if not funcionarios:
        print("\nNenhum funcionário cadastrado.")
        return

    funcionario = max(funcionarios, key=lambda f: f.salario)

    print("\n=== MAIOR SALÁRIO ===")
    print(f"Nome: {funcionario.nome}")
    print(f"Cargo: {funcionario.cargo}")
    print(f"Salário: R$ {funcionario.salario:.2f}")

def media_salarial():
    if not funcionarios:
        print("\nNenhum funcionário cadastrado.")
        return

    media = sum(f.salario for f in funcionarios) / len(funcionarios)

    print(f"\nMédia salarial: R$ {media:.2f}")

def acima_da_media():
    if not funcionarios:
        print("\nNenhum funcionário cadastrado.")
        return

    media = sum(f.salario for f in funcionarios) / len(funcionarios)

    print(f"\nMédia salarial: R$ {media:.2f}")
    print("\nFuncionários acima da média:")

    encontrados = False

    for funcionario in funcionarios:
        if funcionario.salario > media:
            encontrados = True
            print(
                f"- {funcionario.nome} | "
                f"{funcionario.cargo} | "
                f"R$ {funcionario.salario:.2f}"
            )

    if not encontrados:
        print("Nenhum funcionário possui salário acima da média.")

while True:
    print("\n" + "=" * 35)
    print("       SISTEMA DE FUNCIONÁRIOS")
    print("=" * 35)
    print("1 - Cadastrar funcionário")
    print("2 - Listar funcionários")
    print("3 - Maior salário")
    print("4 - Média salarial")
    print("5 - Salários acima da média")
    print("0 - Sair")

    opcao = input("\nEscolha uma opção: ")

    if opcao == "1":
        cadastrar()
    elif opcao == "2":
        listar()
    elif opcao == "3":
        maior_salario()
    elif opcao == "4":
        media_salarial()
    elif opcao == "5":
        acima_da_media()
    elif opcao == "0":
        print("\nPrograma encerrado.")
        break
    else:
        print("\nOpção inválida.")
````
